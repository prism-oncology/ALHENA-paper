# Get an overview of the distribution of dispersions_norm for genes for each adata to help decide the min_disp for hvg selection

**Note**: these plots use all the cell IDs of each adata, whereas gene selection uses only the train cell IDs. The distributions are still indicative.

```python
import sys
from pathlib import Path
import scanpy as sc
import anndata as ad
import matplotlib.pyplot as plt

# Add the project root to sys.path so that 'src' can be imported
project_root = Path.cwd().parents[2]
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.utils.constants import get_full_path
```

```python
def plot_hist_dispersions_norm(slide_id):

    print(f"\n\n\n######## {slide_id} ########\n")

    adata_path = get_full_path('STHELAR_HistoST', 'adata_graph', as_str=True)
    adata = ad.read_h5ad(f"{adata_path}/{slide_id}.h5ad")
    
    if 'dispersions_norm' in adata.var:
        raise ValueError(f"adata.var already contains 'dispersions_norm'.")
    if adata.X.min() < 0:
        raise ValueError(f"Negative values found in adata.X. Are you sure this is raw counts in adata.X?")
    max_value = adata.X.max()
    if max_value < 10:
        raise ValueError(f"Max value in adata.X is {max_value}, which is less than 10. Are you sure this is raw counts in adata.X?")
    if int(max_value) != max_value:
        raise ValueError(f"Max value in adata.X is {max_value}, which is not an integer. Are you sure this is raw counts in adata.X?")
    print(f"[INFO] Normalizing and logarithmizing the data")
    sc.pp.normalize_total(adata, target_sum=1e4)
    sc.pp.log1p(adata) 
    print("[INFO] Computing highly variable genes")
    sc.pp.highly_variable_genes(adata, flavor="seurat")
    
    print(adata)
    print("\n")
    print("Number of genes:", adata.shape[1])
    plt.hist(adata.var['dispersions_norm'], bins=100)
    plt.show()
```

```python
# Get slide_ids_list
dataset_id = "ds_ct1_20x_224_o112"
slide_ids_txt = get_full_path('train', 'datasets', dataset_id, 'slide_ids.txt', as_str=True)
with open(slide_ids_txt, 'r') as f:
    slide_ids_list = [line.strip() for line in f.readlines()]
print(f"slide_ids_list: {slide_ids_list}")
print(f"Number of slides: {len(slide_ids_list)}")
```

```python
for slide_id in slide_ids_list:
    plot_hist_dispersions_norm(slide_id)
```
